Task 1

```python
# iter(apps) list ne iterator ma convert kare chhe.
# next(app_iterator) ek-ek kari ne next item aape chhe.
# List na badha items purા thai jay pachhi vadhu next() call karsho to StopIteration error aavshe.




apps = ["Zomato", "Swiggy", "Dominos", "Zepto", "Blinkit"]

app_iterator = iter(apps)

print(next(app_iterator))
print(next(app_iterator))
print(next(app_iterator))
print(next(app_iterator))
print(next(app_iterator))
```

Task 2

```python
# yield ek vakhat ek song return kare chhe.
# Generator badha songs ek sathe return karvane badle ek-ek song aape chhe.
# Aa example songs ne shuffle kartu nathi; te original order ma songs aape chhe.




def playlist_generator(songs):
    for song in songs:
        yield song


songs = ["Shape Of You", "Blinding Lights", "Levitating", "Senorita"]

playlist = playlist_generator(songs)

for song in playlist:
    print(song)
```

Task 3

```python
# enumerate() darek item sathe teno index pan aape chhe.



cart = ["Pizza", "Burger", "Fries", "Coke"]

for index, item in enumerate(cart):
    print(index, item)
```

OR

```python
# Jo numbering 1 thi start karvi hoy:

for index, item in enumerate(cart, start=1):
    print(index, item)
```

Task 4

```python
# zip(teams, points) banne lists na corresponding items ne pair kare chhe.
# team ma team name ane point ma tena points aave chhe.
# Aa example na points practice mate chhe, actual IPL standings nathi.




teams = ["Mumbai Indians", "Chennai Super Kings",
         "Royal Challengers Bengaluru", "Rajasthan Royals"]

points = [18, 16, 14, 12]

for team, point in zip(teams, points):
    print(f"Team: {team}, Points: {point}")
```

Task 5

```python
# order_id = 1001 thi starting thay chhe.
# yield order_id current ID aape chhe.
# order_id += 1 next ID mate number ek vadhare chhe.
# while True generator ne continuously IDs generate karva de chhe, jyare darek next() call par ek navi ID male chhe.
# Note: Aa demo generator chhe. Real app ma unique order IDs generate karva database athva biji uniqueness system pan jaruri hoy chhe.

def order_id_generator():
    order_id = 1001

    while True:
        yield order_id
        order_id += 1


orders = order_id_generator()

print(next(orders))
print(next(orders))
print(next(orders))
print(next(orders))
print(next(orders))
```

IMPORTANT

```python
# Concept	        --Use
# iter()	        --Iterator banave
# next()	        --Next item aape
# yield	            --Generator mathi ek value aape
# enumerate()	    --Index ane item aape
# zip()	            --Be ke vadhu sequences na items pair kare
# Generator	        --Values ne jarur pade tyare ek-ek kari ne generate kare
```
