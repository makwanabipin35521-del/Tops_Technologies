```python
# 1. songs[0] list nu first song print kare chhe.
# 2. songs[1:] first song vagar ni baki list aape chhe.
# 3. Function baki songs sathe potane fari call kare chhe.
# 4. List empty thay tyare return function ne stop kare chhe.

# Udaharan: ["A", "B", "C"] → "A" print → ["B", "C"] → "B" print → ["C"] → "C" print → empty list, stop.





def print_playlist_songs(songs):
    if len(songs) == 0:
        return

    print(songs[0])
    print_playlist_songs(songs[1:])


songs = ["Shape Of You", "Blinding Lights", "Levitating", "Senorita"]

print_playlist_songs(songs)
```

Task 2

```python
# Main group: 5 messages
# First subgroup: 10 messages
# Ena be subgroups: 3 + 2 messages
# Second subgroup: 7 messages
```

```python
# .get("count", 0) count key na hoy to 0 aape chhe.
# .get("subgroups", []) subgroup na hoy to empty list aape chhe.
# count_unread_messages(group) function ne fari call kare chhe.
# return total badha groups no total aape chhe.



def count_unread_messages(messages):
    total = messages.get("count", 0)

    for group in messages.get("subgroups", []):
        total += count_unread_messages(group)

    return total


whatsapp_groups = {
    "count": 5,
    "subgroups": [
        {
            "count": 10,
            "subgroups": [
                {"count": 3, "subgroups": []},
                {"count": 2, "subgroups": []}
            ]
        },
        {
            "count": 7,
            "subgroups": []
        }
    ]
}

print("Total unread messages:", count_unread_messages(whatsapp_groups))
```

Task 3

```python
# x = 'global' global variable banave chhe.
# outer() call thay tyare local x = 'outer' bane chhe.
# inner() ma nonlocal x outer function na variable ne refer kare chhe.
# x = 'inner' outer function na x ne update kare chhe.
# Etle Inside outer: inner print thay chhe.
# Global x badlatu nathi, etle Outside: global print thay chhe.



x = 'global'

def outer():
    x = 'outer'

    def inner():
        nonlocal x
        x = 'inner'

    inner()
    print('Inside outer:', x)

outer()
print('Outside:', x)
```

```python
# Instagram ke YouTube follower count ne K ane M format ma convert kariye.
# - 1500 → 1.5K
# - 1200000 → 1.2M
# - 500 → 500


def format_number_short(n, suffix_index=0):
    suffixes = ["", "K", "M", "B", "T"]

    if n < 1000 or suffix_index == len(suffixes) - 1:
        if suffix_index == 0:
            return str(n)
        return f"{n:.1f}{suffixes[suffix_index]}"

    return format_number_short(n / 1000, suffix_index + 1)


print(format_number_short(1500))
print(format_number_short(1200000))
print(format_number_short(500))
```
