```python
# Base Class (Parent Class)

class Member:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def monthly_fee(self):
        return 1200

    def describe(self):
        return f"Member Name: {self.name}, Age: {self.age}"


# Child Class 1

class StudentMember(Member):

    def __init__(self, name, age):
        super().__init__(name, age)


# Child Class 2

class PremiumMember(Member):

    def __init__(self, name, age, trainer_name):
        super().__init__(name, age)
        self.trainer_name = trainer_name


# Create Objects

member1 = Member("Bipin", 21)

member2 = StudentMember("Shreya", 18)

member3 = PremiumMember("Amit", 28, "John")



# Print describe() for all three objects

print(member1.describe())
print(member2.describe())
print(member3.describe())
```

```python
# Parent Class

class Member:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def monthly_fee(self):
        return 1200

    def describe(self):
        return f"Member Name: {self.name}, Age: {self.age}"


# Child Class 1

class StudentMember(Member):

    def __init__(self, name, age):
        super().__init__(name, age)

    # Method Overriding
    def monthly_fee(self):
        return super().monthly_fee() * 0.80


# Child Class 2

class PremiumMember(Member):

    def __init__(self, name, age, trainer_name):
        super().__init__(name, age)
        self.trainer_name = trainer_name

    # Method Overriding
    def monthly_fee(self):
        return super().monthly_fee() + 800


# Create Objects

member1 = Member("Bipin", 25)

member2 = StudentMember("Shreya", 18)

member3 = PremiumMember("Rishi", 20, "Yash")


# Store objects in a list

members = [member1, member2, member3]


# Print monthly fees using loop

for m in members:
    print(type(m).__name__, ":", m.monthly_fee())
```

```python
# Parent Class

class Member:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def monthly_fee(self):
        return 1200

    def describe(self):
        return f"Member Name: {self.name}, Age: {self.age}"


# Child Class 1

class StudentMember(Member):

    def __init__(self, name, age):
        super().__init__(name, age)

    def monthly_fee(self):
        return super().monthly_fee() * 0.80


# Child Class 2

class PremiumMember(Member):

    def __init__(self, name, age, trainer_name):
        super().__init__(name, age)
        self.trainer_name = trainer_name

    def monthly_fee(self):
        return super().monthly_fee() + 800


# Function to calculate total monthly income

def total_monthly_income(members):
    total = 0

    for member in members:
        total += member.monthly_fee()

    return total


# Create 8 members

members = [

    Member("Bipin", 21),
    Member("Shreya", 20),
    Member("rishi", 21),

    StudentMember("Bipin", 21),
    StudentMember("Shreya", 18),
    StudentMember("Rishi ", 20),

    PremiumMember("Ayush", 20, "John"),
    PremiumMember("Yash", 21, "David")
]


# Count students using isinstance()
# aa check kare chhe ke object StudentMember class no chhe ke nahi.

student_count = 0

for member in members:
    if isinstance(member, StudentMember):
        student_count += 1


# Calculate income per class
# aa fakt exact Member objects count kare chhe. StudentMember અને PremiumMember ne alag count karva mate aa jarur chhe.
member_count = sum(type(m) is Member for m in members)
student_income_count = sum(isinstance(m, StudentMember) for m in members)
premium_count = sum(isinstance(m, PremiumMember) for m in members)

member_income = sum(
    m.monthly_fee() for m in members if type(m) is Member
)

student_income = sum(
    m.monthly_fee() for m in members if isinstance(m, StudentMember)
)

premium_income = sum(
    m.monthly_fee() for m in members if isinstance(m, PremiumMember)
)


# Calculate total income

total_income = total_monthly_income(members)


# Print student count

print("Total Students:", student_count)

# Print total monthly income

print("Total Monthly Income: ₹", total_income)


# Final table

print("\nClass Name       Count       Income")
print("-------------------------------------")

print(f"Member           {member_count}           ₹{member_income}")
print(f"StudentMember    {student_income_count}           ₹{student_income}")
print(f"PremiumMember    {premium_count}           ₹{premium_income}")

print("-------------------------------------")
print(f"Total            {len(members)}           ₹{total_income}")
```
