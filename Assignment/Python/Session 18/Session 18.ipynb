Task 2

```python
# 0[1-9]|[12]\d|3[01] day માટે 01 થી 31 સુધીના values શોધે છે.
# 0[1-9]|1[0-2] month માટે 01 થી 12 સુધીના values શોધે છે.
# \d{4} ચાર digits નું year શોધે છે.



def check_date(text):
    pattern = r'\b(0[1-9]|[12]\d|3[01])/(0[1-9]|1[0-2])/\d{4}\b'

    if re.search(pattern, text):
        return True
    else:
        return False

print(check_date("My birthday is 18/02/2005"))
print(check_date("My birthday is 18/13/2005"))
```

Task 3

```python


text = "Mobile Rs. 1500, Earphones Rs. 299, Cover Rs. 150"

prices = re.findall(r'Rs\.\s*(\d+)', text)

prices = [int(price) for price in prices]

print("Prices:", prices)
print("Total:", sum(prices))
```

```python


text = "Contact bipin@gmail.com or sagar@yahoo.com for details."

result = re.sub(
    r'\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b',
    '[hidden email]',
    text
)

print(result)
```
