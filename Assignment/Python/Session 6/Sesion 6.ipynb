```python
# .items() dictionary માંથી key અને value બંને આપે છે.
# for દરેક food item પર એક પછી એક ચાલે છે.
# if price > 200 ફક્ત ₹200 કરતાં વધુ કિંમતવાળી items પસંદ કરે છે.

food_prices = {
    "Pizza": 350,
    "Burger": 150,
    "Paneer Tikka": 250,
    "Dosa": 120,
    "Biryani": 280
}

for item, price in food_prices.items():
    if price > 200:
        print(item, "₹", price)
```

```python
# set unique values store કરે છે.
# .intersection() બંને sets માં common હોય તેવા usernames આપે છે.
# & operator વડે પણ intersection કરી શકાય છે.

flipkart_users = {
    "nayan", "suraj", "bipin", "mukesh", "rahul"
}

myntra_users = {
    "bipin", "rahul", "amit", "jay", "nayan"
}

common_users = flipkart_users.intersection(myntra_users)

print(common_users)
```

```python
#def function બનાવવા માટે વપરાય છે.
# union() બંને sets ના બધા unique artists આપે છે.
# બંને playlists માં એક જ artist હોય તો પણ તે result માં એક જ વાર આવશે.
# return result પાછું આપે છે.

def get_unique_artists(spotify_playlist1, spotify_playlist2):
    return spotify_playlist1.union(spotify_playlist2)


spotify_playlist1 = {"Arijit Singh", "Shreya Ghoshal", "Atif Aslam"}
spotify_playlist2 = {"Atif Aslam", "Sonu Nigam", "Arijit Singh"}

result = get_unique_artists(spotify_playlist1, spotify_playlist2)

print(result)
```
