Task 1

```python
# lambda song: song.lower() darek song ne lowercase kare chhe.
# map() list na darek song par function apply kare chhe.
# list() result ne list ma convert kare chhe.



songs = ['Shape Of You', 'Blinding Lights', 'Levitating', 'Senorita']

result = list(map(lambda song: song.lower(), songs))

print(result)
```

Task 2

```python
# filter() khali e ratings select kare chhe je 4.0 karta vadhu chhe.



ratings = [4.2, 3.8, 4.5, 2.9, 3.5]

result = list(filter(lambda rating: rating > 4.0, ratings))

print(result)
```

Task 3

```python
# reduce() be-be values ne combine kare chhe.
# lambda x, y: x + y values no sum kare chhe.
# Total = 499 + 1299 + 299 + 799 = ₹2896.



from functools import reduce

prices = [499, 1299, 299, 799]

total = reduce(lambda x, y: x + y, prices)

print("Total price: ₹", total)
```

Task 4

```python
# 1000 ke vadhu hoy to K format.
# 1000000 ke vadhu hoy to M format.
# round(..., 1) decimal pachhi ek digit rakhe chhe.
# map() function ne darek follower count par apply kare chhe.




def format_followers(count):
    if count >= 1000000:
        return str(round(count / 1000000, 1)) + "M"
    elif count >= 1000:
        return str(round(count / 1000, 1)) + "K"
    else:
        return str(count)


followers = [950, 1500, 25000, 1200000]

result = list(map(format_followers, followers))

print(result)
```

Task 5

```python
# score % 2 == 0 check kare chhe ke number even chhe ke nahi.
# filter() khali even numbers rakhe chhe, etle badha odd numbers remove thay chhe.



scores = [101, 98, 120, 77, 88]

result = list(filter(lambda score: score % 2 == 0, scores))

print(result)
```
