```python
class Member:

    def __init__(self, name, age, plan):
        self.name = name
        self.age = age
        self.plan = plan
        self.visits = []  # Each object gets its own list

    # Add gym visit date
    def check_in(self, date_text):
        self.visits.append(date_text)

    # Count total visits
    def visit_count(self):
        return len(self.visits)

    # Return  price
    def plan_price(self):

        if self.plan == "Monthly":
            return 1200

        elif self.plan == "Quarterly":
            return 3300

        elif self.plan == "Annual":
            return 11000


# Creating five objects

member1 = Member("Rahul", 22, "Monthly")
member2 = Member("Priya", 25, "Quarterly")
member3 = Member("Amit", 28, "Annual")
member4 = Member("Neha", 21, "Monthly")
member5 = Member("Karan", 30, "Annual")


# Check-in dates for first member
member1.check_in("2026-03-01")
member1.check_in("2026-03-03")
member1.check_in("2026-03-05")


# Check-in dates for second member
member2.check_in("2026-03-02")
member2.check_in("2026-03-04")

# third member not checked in at gym
# fourth member not checked in at gym
# fifth member not checked in at gym

# Store all members in a list
members = [member1, member2, member3, member4, member5]


# Print visit count and plan price
for i in members:
    print("Name:", i.name)
    print("Visit Count:", i.visit_count())
    print("Plan Price: ₹", i.plan_price())
    print("----------------------")
```

```python
class Member:

    def __init__(self, name, age, plan):
        self.name = name
        self.age = age
        self.plan = plan
        self.visits = []

    # Add visit date 
    # upar add karya temj
    def check_in(self, date_text):
        self.visits.append(date_text)

    # Count visits
    def visit_count(self):
        return len(self.visits)

    # Calculate plan price
    def plan_price(self):

        if self.plan == "Monthly":
            return 1200

        elif self.plan == "Quarterly":
            return 3300

        elif self.plan == "Annual":
            return 11000


# Create six members

member1 = Member("Bipin", 22, "Monthly")
member2 = Member("Rishi", 25, "Monthly")
member3 = Member("Shreya", 28, "Quarterly")
member4 = Member("Kaveni", 21, "Annual")
member5 = Member("Yash", 30, "Quarterly")
member6 = Member("Karan", 24, "Monthly")


# Add check-in dates

member1.check_in("2026-03-01")
member1.check_in("2026-03-03")
member1.check_in("2026-03-05")

member2.check_in("2026-03-02")

member3.check_in("2026-03-01")
member3.check_in("2026-03-04")

member4.check_in("2026-03-02")
member4.check_in("2026-03-05")
member4.check_in("2026-03-07")
member4.check_in("2026-03-09")

member5.check_in("2026-03-03")

member6.check_in("2026-03-01")
member6.check_in("2026-03-04")


# Store all members in a list

members = [member1, member2, member3,member4, member5, member6]


# Calculate total revenue

total = sum(member.plan_price() for member in members)


# Find member with most visits
# i use lambda function to find the member with the maximum visit count

top_member = max(members, key=lambda member: member.visit_count())


# Create report

report = (
    f"Total Revenue: Rs. {total}\n"
    f"Member with most visits: {top_member.name}\n"
)



print(report)


# Write report into text file
# all report will be save in gym_report.txt.

with open("gym_report.txt", "w") as file:
    file.write(report)

print("Report saved successfully!")
```
