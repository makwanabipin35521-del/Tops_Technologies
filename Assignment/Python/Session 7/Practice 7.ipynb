```python
#marks માં students ના marks છે.
# if mark >= 35 એટલે 35 કે તેથી વધુ marks મેળવનાર student પાસ.
# passed += 1 એટલે પાસ થયેલા વિદ્યાર્થીઓની ગણતરીમાં 1 ઉમેરવો.
marks= [78, 25, 45, 32, 90, 35, 60]

passed = 0

for mark in marks:
    if mark >= 35:
        passed += 1

print("Students passed:", passed)
```

```python
# 13 * i multiplication નું result આપે છે.
# ["13"] * i string "13" ને i વખત list માં મૂકે છે.
# "+".join(...) બધા "13" વચ્ચે + મૂકે છે.

for i in range(1, 11):
    result = 13 * i
    repeated = "+".join(["13"] * i)

    print(f"13*{i}=({repeated})={result}")
```

```python
# highest_sale = sales[0] શરૂઆતમાં પહેલી sale ને highest માને છે.
# for sale in sales દરેક sale ચેક કરે છે.
# જો કોઈ sale હાલની highest sale કરતાં મોટી હોય, તો highest_sale update થાય છે.

sales = [1200, 2500, 1800, 4500, 3200]

highest_sale = sales[0]

for sale in sales:
    if sale > highest_sale:
        highest_sale = sale

print("Highest sale:", highest_sale)
```
