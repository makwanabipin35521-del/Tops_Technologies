Task 2

```python
# append(106) → એક song ID ઉમેરે છે.
# extend([107]) → બીજી list ના બધા elements ઉમેરે છે.

playlist_ids = [101, 102, 103, 104, 105]
playlist_ids2=[107,108,109]

# Add a single song ID
playlist_ids.append(106)

# Add multiple song IDs
playlist_ids.extend(playlist_ids2)

print(playlist_ids)
```

Task 3

```python
# pop() list નો છેલ્લો item દૂર કરે છે.
# દૂર કરેલી value removed_id માં store થાય છે.
# પછી બાકીની playlist print થાય છે.

playlist_ids = [101, 102, 103, 104, 105]

removed_id = playlist_ids.pop()

print("Removed Song ID:", removed_id)
print("Remaining Playlist:", playlist_ids)
```

Task 4

```python
# Tuple Ma Value Change Nahi Thaay Ane Error Aavse
# tuple માં square brackets [] ના બદલે parentheses () વપરાય છે.
# tuple માં values immutable હોય છે એટલે value change ન કરી શકાય.

insta_filters = ("Clarendon", "Juno", "Lark", "Gingham")

print(insta_filters)

insta_filters[0] = "Vintage"
```

Task 5

```python
# List: જ્યારે data add, remove અથવા update કરવો હોય ત્યારે ઉપયોગી છે.
# Tuple: જ્યારે data ને સ્થિર રાખવો હોય અને તેની values બદલવાની જરૂર ન હોય ત્યારે ઉપયોગી છે.

# Recent Zomato orders can change, so use a list.
zomato_orders = ["Pizza", "Burger", "Dosa"]
print("Zomato orders (list):", zomato_orders)

# Fixed IPL team names should not change, so use a tuple.
ipl_teams = ("CSK", "MI", "RCB", "KKR")
print("IPL teams (tuple):", ipl_teams)
```

Important Note

```python
# Method / Concept	ઉપયોગ
# append()	એક item ઉમેરવો
# extend()	બીજા iterable માંથી અનેક items ઉમેરવા
# pop()	Item દૂર કરીને તેની value મેળવવી
# list []	Mutable collection
# tuple ()	Immutable collection
# યાદ રાખો: List બદલાઈ શકે છે, જ્યારે Tuple ના elements બદલી શકાતા નથી.
```
