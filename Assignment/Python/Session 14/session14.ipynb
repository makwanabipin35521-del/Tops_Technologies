```python
# open() = opens or creates a file.
# "w" = means write mode.
# \n = moves the next song to a new line.
# with = automatically closes the file when the operation finishes


with open("playlist.txt", "w") as file:
    file.write("Perfect\n")
    file.write("Shape of You\n")
    file.write("Believer\n")
    file.write("Kesariya\n")
    file.write("Heeriye\n")

print("Songs saved successfully!")
```

Task 2

```python
# "r" =              means read mode.
# for song in file = reads the file line by line.
# strip() =          removes extra whitespace and the newline character.
# upper() =          converts the text to uppercase.

with open("playlist.txt", "r") as file:
    for song in file:
        print(song.strip().upper())
```

```python
# ipl.csv file banaavi pachhi tema 4 data add karya chhe.
import csv
with open("ipl.csv","w",newline="") as file:
    writer=csv.writer(file)
    writer.writerow(["match_id","team1","team2","winner"])
    writer.writerow(["CSK","MI","CSK"])
    writer.writerow(["RCB","KKR","KKR"])
    writer.writerow(["GT","RR","GT"])
    writer.writerow(["MI","RCB","RCB"])
    
```

```python
# badha data print karaya.
import csv
with open("ipl.csv","r") as file:
    reader=csv.reader(file)
    for i in reader:
        print(i)
        
```

```python
# winner show karse
import csv

with open("ipl.csv", "r") as file:
    reader = csv.DictReader(file)

    for match in reader:
        print("Match", match["match_id"], "- Winner:",match["winner"])
```

```python
# json file ma 1 dictionary ma 3 data add karya. 
movie=[
    {
        "title": "3 Idiots","year": 2009,"rating": 8.4},
    {
        "title": "Dangal","year": 2016,"rating": 8.3},
    {
        "title": "12th Fail","year": 2023,"rating": 8.8
    }
]
```

```python
# movie1.json file create kari.
import json

with open("Movie1.json", "w") as file:
    json.dump(movie,file)
```

```python
# - import json imports the JSON module.
# - json.load(file) reads JSON data and converts it into Python objects.
# - for movie in movies loops through each movie.
# - movie["title"] and movie["rating"] access individual values.
import json

with open("Movie1.json", "r") as file:
    movies = json.load(file)

for movie in movies:
    print("Title:", movie["title"])
    print("Rating:", movie["rating"])
    print()
```

```python
import json
import os

file_name = "my_fav_apps.json"

apps = [
    {"name": "Instagram", "category": "Social Media"},
    {"name": "Zomato", "category": "Food Delivery"},
    {"name": "Paytm", "category": "Finance"}
]

if not os.path.exists(file_name):
    with open(file_name, "w") as file:
        json.dump(apps, file, indent=4)

    print("File created successfully!")
else:
    print("File already exists!")
```

```python
import json
with open("my_fav_apps.json","r") as file:
    reader=json.load(file)
    for i in reader:
        print(i)
```
