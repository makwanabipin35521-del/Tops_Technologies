task 1

```python
# lower() → બધા letters ને lowercase કરે છે.
# replace('-', ' ') → dash (-) ને space માં બદલે છે.
text = 'Flipkart-Sale2024'

result = text.lower().replace('-', ' ')

print(result)
```

task 2

```python
# strip() → શરૂઆત અને અંતમાં રહેલી extra spaces દૂર કરે છે.
# upper() → બધા letters ને CAPITAL બનાવે છે.
# replace('-', ':') → dash ને colon (:) માં બદલે છે.

product = ' OnePlus Nord-CE 3 '
result = product.strip().upper().replace('-', ':')

print(result)
```

Task 3

```python
# def → function બનાવવા માટે વપરાય છે.
# split_product_code(product_code) → function નું નામ અને parameter છે.
# split('-') → dash જ્યાં હોય ત્યાં string ને અલગ parts માં વહેંચે છે.
# return → result પાછું આપે છે.
# અહીં result એક list છે.

def split_product_code(product_code):
    return product_code.split("-")


result = split_product_code("ZOMATO-FOOD-2024")

print(result)


```

```python
# f'...' → string માં variables ની values મૂકવા માટે વપરાય છે.
# {product} → Myntra Shirt print કરશે.
# {price:.2f} → price ને decimal પછી 2 digits સાથે print કરશે.
# 799.5 → 799.50 બનીને print થશે.

product = 'Myntra Shirt'
price = 799.5

print(f'Deal: {product} is available at ₹{price:.2f} only!')
```
