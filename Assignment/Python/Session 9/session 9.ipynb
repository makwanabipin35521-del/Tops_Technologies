Task 1

```python
def calculate_final_price(price, discount_rate):
    discount = price * discount_rate / 100
    final_price = price - discount
    return final_price


result = calculate_final_price(1000, 20)
print("Final price: ₹", result)
```

```python
# Default currency INR chhe.
# str(price) number ne string ma convert kare chhe, jethi currency symbol sathe jodi shakay.
# USD aapo to $ symbol return thase.


def format_price(price, currency="INR"):
    if currency == "INR":
        return "₹" + str(price)
    elif currency == "USD":
        return "$" + str(price)


print(format_price(500))
print(format_price(500, "USD"))
```

Task 4

```python
# coupon_code=None default argument chhe.
# Coupon "ZOMATO10" hoy to 10% discount male.
# Coupon na aapo athva invalid coupon aapo to original price return thase.



def apply_coupon(price, coupon_code=None):
    if coupon_code == "ZOMATO10":
        return price - (price * 10 / 100)
    else:
        return price


print(apply_coupon(500, "ZOMATO10"))
print(apply_coupon(500))
print(apply_coupon(500, "INVALID"))
```
