Task 2

```python
# while loop darek divas na steps check kare chhe.
# steps[i] > 10000 condition check kare chhe.
# break pehli vaar 10,000 karta vadhu steps male tyare loop bandh kare chhe.
# i += 1 next day par lai jay chhe.



steps = [4500, 7800, 9500, 12000, 8500, 11000, 6000]
days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

i = 0

while i < len(steps):
    if steps[i] > 10000:
        print("Day:", days[i])
        print("Steps:", steps[i])
        break
    i += 1
```

Task 3

```python
#len(team) team na characters count kare chhe.
# continue 6 ke tena karta ochha characters hoy to team skip kare chhe.
# Baki team na names print thay chhe.




def print_long_teams(teams):
    for team in teams:
        if len(team) <= 6:
            continue
        print(team)


ipl_teams = [
    "CSK", "MI", "RCB", "Kolkata",
    "Rajasthan", "Hyderabad", "Punjab"
]

print_long_teams(ipl_teams)
```

Task 4

```python
# enumerate() position ane value banne aape chhe.
# start=1 numbering 1 thi sharu kare chhe.
# # f"..." string ni andar variables ni values print kare chhe.



song_durations = [210, 185, 240, 195, 300]

for position, duration in enumerate(song_durations, start=1):
    print(f"Song {position}: {duration} seconds")
```

Task 5

```python
# - continue → ₹0 ni item skip kare chhe.
# - break → next item add karta total ₹2000 thi vadhu thay to loop stop kare chhe.
# - total += price → current price total ma add kare chhe.
# Ahiya ₹500 + ₹300 + ₹700 = ₹1500 thay chhe. Pachhi ₹800 add karvathi total ₹2300 thai jay, etle break loop bandh kare chhe.




prices = [500, 0, 300, 700, 800, 200]

total = 0

for price in prices:
    if price == 0:
        continue

    if total + price > 2000:
        break

    total += price

print("Final total: ₹", total)
```

IMPORTTANT

```python
"""Concept	Use
for	List na elements par loop chalavvo
while	Condition true hoy tya sudhi loop chalavvo
break	Loop ne bandh karvo
continue	Current item skip karvo
enumerate()	Index/position ane value banne melavva
len()	Length ke number of items janva"""
```
