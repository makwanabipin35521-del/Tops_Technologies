```python
ipl_teams = ["CSK","MI","RCB","Kolkata","Rajasthan","Hyderabad","Punjab"]
def print_long_teams(teams):
    for team in teams:
        if len(team) <= 6:
            continue
        print(team)


ipl_teams = [
    "CSK",
    "MI",
    "RCB",
    "Kolkata",
    "Rajasthan",
    "Hyderabad",
    "Punjab"
]

print_long_teams(ipl_teams)
```
