```python
# gym_encapsulation.ipynb

class Member:

    def __init__(self, name, age, plan):
        self.name = name
        self.__age = age
        self.plan = plan
        self.visits = []

        # Set fee according to membership plan
        if self.plan == "Monthly":
            self.__fee_due = 1200

        elif self.plan == "Quarterly":
            self.__fee_due = 3300

        elif self.plan == "Annual":
            self.__fee_due = 11000

        else:
            raise ValueError("Invalid membership plan") # exception handling

    # Getter for age
    def get_age(self):
        return self.__age

    # Setter for age
    def set_age(self, new_age):
        if 14 <= new_age <= 80:
            self.__age = new_age
        else:
            raise ValueError("Age must be between 14 and 80")

    # Payment method
    def pay(self, amount):

        if amount <= 0:
            raise ValueError("Payment must be greater than 0")

        elif amount > self.__fee_due:
            raise ValueError("Payment cannot exceed the balance due")

        else:
            self.__fee_due -= amount
            print("Payment successful:", amount)

    # Check remaining balance
    def balance_due(self):
        return self.__fee_due


# Create Quarterly member
member = Member("Rahul", 25, "Quarterly")

# Initial balance
print("Initial Balance:", member.balance_due())


# First payment
member.pay(1000)
print("Balance after first payment:", member.balance_due())


# Second payment
member.pay(2300)
print("Balance after second payment:", member.balance_due())


# Try accessing private variable directly
try:
    print(member.__fee_due)

except AttributeError as e:
    print("Error:", e)


# Access using name mangling
print("Private Fee Due:", member._Member__fee_due)


# i use this topics:

#- Encapsulation
#- Private variables
#- Getter and Setter
#- ValueError
#- Payment validation
#- Name Mangling
#- Balance calculation
```

```python
from datetime import datetime, timedelta


class Member:

    def __init__(self, name, age, plan):
        self.name = name
        self.__age = age
        self.plan = plan
        self.visits = []


class GymDesk:

    # Private method 1: Calculate renewal amount
    def __calculate_amount(self, member, months):

        amount = 1200 * months

        if member.plan == "Monthly" and months >= 6:
            amount = amount - (amount * 0.10)

        return amount

    # Private method 2: Calculate valid-till date
    def __calculate_valid_till(self, months):

        start_date = datetime(2026, 4, 1)

        valid_date = start_date + timedelta(days=30 * months)

        return valid_date.strftime("%Y-%m-%d")

    # Only public method
    def renew_membership(self, member, months):

        amount = self.__calculate_amount(member, months)

        valid_till = self.__calculate_valid_till(months)

        receipt = {
            "Member Name": member.name,
            "Plan": member.plan,
            "Months": months,
            "Amount": amount,
            "Valid Till": valid_till
        }

        return receipt


# Create Monthly member
member = Member("Bipin", 25, "Monthly")

# Create  object
desk = GymDesk()


# Renew membership for 1, 3 and 6 months

receipt1 = desk.renew_membership(member, 1)
receipt2 = desk.renew_membership(member, 3)
receipt3 = desk.renew_membership(member, 6)


# Print receipts

print("1 Month Receipt:")
print(receipt1)

print("\n3 Months Receipt:")
print(receipt2)

print("\n6 Months Receipt:")
print(receipt3)
```
