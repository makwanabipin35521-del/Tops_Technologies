Error & Exception handling

```python
try:
    new=int(input("enter no"))
    result= 100/new
    print(result)
except ZeroDivisionError:
    print("you can't divide")
except ValueError:
    print("Please enter integer value")
else:
    print("divition succesfully")
finally:
    print("i did it")


```
