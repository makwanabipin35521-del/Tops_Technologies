Task 1

```python
# - try: Executes the division.
# - except: Handles division by zero.
# - finally: Executes whether aexception occurs or not.


def get_song_duration_per_minute(duration, songs):
    try:
        average = duration / songs
        print(f"Average song duration:{ average } minutes")

    except ZeroDivisionError:
        print("Number of songs cannot be zero.")

    finally:
        print("Spotify playlist calculation done.")


get_song_duration_per_minute(120, 20)
get_song_duration_per_minute(120, 0)
```

```python
try:
    amount = float(input("Enter total cart amount: "))
    count = int(input("Enter number of items: "))

    price = amount / count

    print("Price per item:", price)

# handles a zero item count
except ZeroDivisionError:
    print("Error: Item count cannot be zero.")
# handles invalid input
except ValueError:
    print("Error: Please enter valid numbers.")
```

Task 3

```python
# class — નવી class બનાવવા માટેનો keyword.
# NoOffersApplied — આપણી custom exception class નું નામ.
# (Exception) — Python ની built-in Exception class માંથી નવી class બનાવીએ છીએ.
# pass — અત્યારે class ની અંદર કોઈ વધારાનો code લખવાનો નથી.

class NoOffersApplied(Exception):
    pass

try:
    spend = float(input("Enter total spend: "))
    offers = int(input("Enter number of offers applied: "))

    if offers == 0:
        raise NoOffersApplied("No offers have been applied.")

    cashback = spend / offers
    print("Average cashback per offer:", cashback)

except NoOffersApplied as e:
    print("Error:", e)

except ValueError:
    print("Please enter valid numbers.")
```

```python


def calculate(rating,reviews):
    return rating / reviews


try:
    print(calculate(500, 0))

except ZeroDivisionError:
    print("Number of reviews cannot be zero.")

except Exception as e:
    print("unexpected error occurred:", e)

finally:
    print("Thank you for using the calculator")
```

```python
def zomato(amount, people):
    try:
        total = amount / people

    except ZeroDivisionError:
        print("Error: Number of people cannot be zero.")

    else:
        print("Each person should pay:", total)

    finally:
        print("Split calculation is done")


zomato(1200, 4)
zomato(1200, 0)
```
